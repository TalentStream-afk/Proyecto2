```python
import matplotlib.pyplot as plt
import numpy as np

# 20 notas de estudiantes entre 1 y 10
notas = [5, 7, 8, 4, 6, 9, 10, 5, 7, 8, 6, 5, 4, 9, 8, 7, 6, 5, 8, 9]

# Dibujar el histograma
plt.hist(notas, color='green')
plt.title("Distribución de Notas")
plt.xlabel("Notas")
plt.ylabel("Cantidad de estudiantes")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Datos
frutas = ['Manzanas', 'Peras', 'Bananos']
cantidad = [10, 15, 7]

# Dibujar barras horizontales (barh)
plt.barh(frutas, cantidad, color='orange')
plt.title("Inventario de Frutas")
plt.show()
```
